```python
from transformers import pipeline, AutoTokenizer, AutoModelForSequenceClassification

# 모델 및 토크나이저 이름
model_name = "JunHwi/kold_binary"
tokenizer_name = "monologg/koelectra-base-v3-discriminator"

# 모델 및 토크나이저 로드
tokenizer = AutoTokenizer.from_pretrained(tokenizer_name)
model = AutoModelForSequenceClassification.from_pretrained(model_name)

# 파이프라인 구성
pipe = pipeline("text-classification", model=model, tokenizer=tokenizer)

# 예제 문장 테스트
example = "너 정말 쓸모없는 인간이야."
result = pipe(example)

# 출력 포맷 정리
labels = {0: "not_hate_speech", 1: "hate_speech"}
for r in result:
    label_id = int(r['label'].split('_')[-1]) if 'label' in r else -1
    print(f"예측 결과: {labels.get(label_id, 'Unknown')} ({r['score']:.2%} 확률)")
```

```python
print(pipe(["넌 정말 짜증나"]))
```

```python
print(pipe(["오늘 날씨가 좋네"]))
```
